```python
import networkx as nx
from pyvis.network import Network

G = nx.DiGraph()

G.add_node("Cris", type="Person")
G.add_node("Python", type="Technology")
G.add_node("NetworkX", type="Library")
G.add_node("PyVis", type="Library")

G.add_edge("Cris", "Python", relation="uses")
G.add_edge("Python", "NetworkX", relation="library")
G.add_edge("Python", "PyVis", relation="library")

net = Network(
    height="750px",
    width="100%",
    directed=True,
    bgcolor="#222222",
    font_color="white"
)

net.from_nx(G)

net.write_html("graph.html")
```
